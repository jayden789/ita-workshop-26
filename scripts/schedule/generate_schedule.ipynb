```python
import pandas as pd
import shutil
```

```python
in_file_path = 'ITA 2025 Schedule - Main.csv'
df = pd.read_csv(in_file_path)
shutil.copy2(in_file_path, "../../backend/api/ITA_2025_Schedule.csv")
```

```python
def float_2_int(flo):
    if flo == "":
        return flo
    else:
        return int(flo)
```

```python
def bool_2_str(boo):
    if boo == "":
        return boo
    else:
        return str(boo)
```

```python
df = df.rename(columns={
                    'Last name': 'lastName', 
                    'First name': 'firstName', 
                    'Title': 'title',
                    'Student?': 'student',
                    'Invited?': 'inviteType',
                    'Status': 'status',
                    'Regn. #': 'registration',
                    'Days': 'days',
                    'Title.1': 'paperTitle',
                    'Abstract': 'abstract',
                    'Scheduling comments': 'schedulingComments',
                    'Topic comments': 'topicComments',
                    'Rough Topic': 'roughTopic',
                    'Scheduled': 'scheduled',
                    'Metadata': 'metadata',
                    'Unnamed: 28': ''
                })
df = df.fillna('')
df['registration'] = df['registration'].apply(float_2_int)
df['Paid amount'] = df['Paid amount'].apply(float_2_int)
df['student'] = df['student'].apply(bool_2_str)
```

```python
out_file_path = '../../frontend/src/2025/views/SchedulePage/schedule-2025.json'
df.to_json(out_file_path, orient='records', indent=4)
```
